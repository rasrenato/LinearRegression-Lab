## 3.3 Vector Creation


Data creation routines in NumPy will generally have a first parameter which is the shape of the object. This can either be a single value for a 1-D result or a tuple (n,m,...) specifying the shape of the result. Below are examples of creating vectors using these routines.

```python
# Código baseado no exercício original, adicionando prints para exibir valores, shape e tipo de dados

import numpy as np

# NumPy routines which allocate memory and fill arrays with value

# Vetor de 4 elementos preenchido com zeros
a = np.zeros(4)
print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")

# Vetor de 4 elementos preenchido com zeros (usando tupla para especificar shape)
a = np.zeros((4,))
print(f"np.zeros((4,)) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")

# Vetor de 4 elementos com valores aleatórios entre 0 e 1
a = np.random.random_sample(4)
print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

Some data creation routines do not take a shape tuple:

```python
# Código baseado no exercício original, adicionando prints para exibir valores, shape e tipo de dados

import numpy as np

# NumPy routines which allocate memory and fill arrays with value but do not accept shape as input argument

# Vetor de valores sequenciais de 0 a 3 (arange)
a = np.arange(4.)
print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")

# Vetor de 4 elementos aleatórios entre 0 e 1 (rand)
a = np.random.rand(4)
print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

values can be specified manually as well. 

```python
a = np.array([5,4,3,2])  
print(f"np.array([5,4,3,2]):  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2])  
print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

These have all created a one-dimensional vector  `a` with four elements. `a.shape` returns the dimensions. Here we see a.shape = `(4,)` indicating a 1-d array with 4 elements.

### 3.4.1 Indexing
Elements of vectors can be accessed via indexing and slicing. NumPy provides a very complete set of indexing and slicing capabilities. We will explore only the basics needed for the course here. Reference [Slicing and Indexing](https://NumPy.org/doc/stable/reference/arrays.indexing.html) for more details.

```python
# Código demonstrando operações de indexação em vetores 1D com NumPy

import numpy as np

# Criando um vetor com valores de 0 a 9
a = np.arange(10)
print("Array a:", a)

# Acessando um elemento específico (terceiro elemento, índice 2)
print(f"a[2].shape: {np.shape(a[2])}, a[2] = {a[2]}, Accessing an element returns a scalar")

# Acessando o último elemento usando índice negativo
print(f"a[-1] = {a[-1]}")

# Tentando acessar um índice fora do intervalo e capturando o erro
try:
    c = a[10]  # Índice fora do intervalo, gera um erro
except Exception as e:
    print("The error message you’ll see is:")
    print(e)
```

### 3.4.2 Slicing
Slicing creates an array of indices using a set of three values (`start:stop:step`). A subset of values is also valid. Its use is best explained by example:

```python
# Código atualizado para operações de slicing (fatiamento) em vetores NumPy

import numpy as np

# Criando um vetor de 0 a 9
a = np.arange(10)
print(f"a = {a}")

# Acessando 5 elementos consecutivos (start:stop:step)
c = a[2:7:1]
print(f"a[2:7:1] = {c}")

# Acessando 3 elementos separados por dois índices
c = a[2:7:2]
print(f"a[2:7:2] = {c}")

# Acessando todos os elementos a partir do índice 3
c = a[3:]
print(f"a[3:] = {c}")

# Acessando todos os elementos antes do índice 3
c = a[:3]
print(f"a[:3] = {c}")

# Acessando todos os elementos do vetor
c = a[:]
print(f"a[:] = {c}")
```

### 3.4.6 Vector Vector dot product
The dot product is a mainstay of Linear Algebra and NumPy. This is an operation used extensively in this course and should be well understood. The dot product is shown below.

<img src="./images/C1_W2_Lab04_dot_notrans.gif" width=800> 

The dot product multiplies the values in two vectors element-wise and then sums the result.
Vector dot product requires the dimensions of the two vectors to be the same. 

Let's implement our own version of the dot product below:

**Using a for loop**, implement a function which returns the dot product of two vectors. The function to return given inputs $a$ and $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Assume both `a` and `b` are the same shape.

```python
# Implementação da função para calcular o produto escalar usando um loop for

import numpy as np

def my_dot(a, b):
    """
    Compute the dot product of two vectors

    Args:
        a (ndarray (n,)): input vector
        b (ndarray (n,)): input vector with the same dimension as a

    Returns:
        x (scalar): dot product result
    """
    x = 0  # Inicializa a soma do produto escalar
    for i in range(a.shape[0]):  # Itera sobre os elementos do vetor
        x += a[i] * b[i]  # Soma dos produtos elemento a elemento
    return x

```

```python
# Testando a função com dois vetores 1D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])

# Chamando a função e imprimindo o resultado
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

Note, the dot product is expected to return a scalar value. 

Let's try the same operations using `np.dot`.  

```python
# Melhorando o teste do produto escalar usando np.dot

import numpy as np

# Criando vetores 1D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])

# Calculando o produto escalar usando np.dot
c1 = np.dot(a, b)
c2 = np.dot(b, a)  # Deve ser o mesmo valor pois o produto escalar é comutativo

# Exibindo os resultados de forma mais organizada
print(f"NumPy 1-D np.dot(a, b) = {c1}, shape = {np.shape(c1)}")
print(f"NumPy 1-D np.dot(b, a) = {c2}, shape = {np.shape(c2)}")

# Verificando se o produto escalar é comutativo
if c1 == c2:
    print("✅ Produto escalar é comutativo: np.dot(a, b) == np.dot(b, a)")
else:
    print("⚠️ Produto escalar não é comutativo, algo está errado!")
```

Above, you will note that the results for 1-D matched our implementation.

### 3.4.7 The Need for Speed: vector vs for loop
We utilized the NumPy  library because it improves speed memory efficiency. Let's demonstrate:

```python
# Comparação de desempenho entre np.dot e a implementação com loop for

import numpy as np
import time

# Definindo a semente para reprodutibilidade
np.random.seed(1)

# Criando grandes arrays aleatórios
a = np.random.rand(10_000_000)
b = np.random.rand(10_000_000)

# Medindo o tempo de execução para np.dot (versão vetorizada)
tic = time.time()  # Tempo inicial
c = np.dot(a, b)
toc = time.time()  # Tempo final

# Exibindo resultado e tempo
print(f"np.dot(a, b) = {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms")

# Função my_dot (produto escalar com loop for)
def my_dot(a, b):
    """
    Compute the dot product of two vectors using a loop.

    Args:
        a (ndarray): input vector
        b (ndarray): input vector with same dimension as a

    Returns:
        x (scalar): dot product result
    """
    x = 0
    for i in range(a.shape[0]):
        x += a[i] * b[i]
    return x

# Medindo o tempo de execução para a implementação com loop
tic = time.time()  # Tempo inicial
c = my_dot(a, b)
toc = time.time()  # Tempo final

# Exibindo resultado e tempo
print(f"my_dot(a, b) = {c:.4f}")
print(f"Loop version duration: {1000*(toc-tic):.4f} ms")

# Removendo grandes arrays da memória
del a, b
```

So, vectorization provides a large speed up in this example. This is because NumPy makes better use of available data parallelism in the underlying hardware. GPU's and modern CPU's implement Single Instruction, Multiple Data (SIMD) pipelines allowing multiple operations to be issued in parallel. This is critical in Machine Learning where the data sets are often very large.

```python
import numpy as np
import time
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Criando diferentes arrays para testar regras de broadcasting
test_cases = [
    {"A": np.array([[1, 2, 3], [4, 5, 6]]), "B": np.array([1, 2, 3]), "expected_shape": (2, 3)},  # (2,3) + (1,3) -> (2,3)
    {"A": np.array([[1], [2], [3]]), "B": np.array([1, 2, 3]), "expected_shape": (3, 3)},  # (3,1) + (1,3) -> (3,3)
    {"A": np.array([[1, 2], [3, 4]]), "B": np.array([[10], [20]]), "expected_shape": (2, 2)},  # (2,2) + (2,1) -> (2,2)
    {"A": np.array([1, 2, 3]), "B": np.array([[1], [2], [3]]), "expected_shape": (3, 3)},  # (3,) + (3,1) -> (3,3)
]

results = []

# Testando as regras de broadcasting e medindo o tempo de execução
for test in test_cases:
    A, B, expected_shape = test["A"], test["B"], test["expected_shape"]
    start_time = time.time()
    
    try:
        C = A + B  # Aplicando broadcasting
        success = C.shape == expected_shape  # Verifica se o shape é o esperado
        error_message = None
    except Exception as e:
        C = None
        success = False
        error_message = str(e)
    
    end_time = time.time()
    execution_time = (end_time - start_time) * 1000  # Tempo em milissegundos
    
    results.append({
        "A shape": A.shape,
        "B shape": B.shape,
        "Expected shape": expected_shape,
        "Result shape": C.shape if C is not None else "Error",
        "Success": success,
        "Time (ms)": round(execution_time, 4),
        "Error Message": error_message,
    })

# Criando um DataFrame para visualizar os resultados
df_results = pd.DataFrame(results)

# Exibindo os resultados no notebook
display(df_results)

# Criando gráfico para visualizar os tempos de execução
plt.figure(figsize=(10, 5))
plt.bar(df_results["A shape"].astype(str) + " + " + df_results["B shape"].astype(str), df_results["Time (ms)"])
plt.xlabel("Operação de Broadcasting")
plt.ylabel("Tempo de Execução (ms)")
plt.title("Performance de Broadcasting em NumPy")
plt.xticks(rotation=45, ha='right')

# Exibir gráfico
plt.show()
```
